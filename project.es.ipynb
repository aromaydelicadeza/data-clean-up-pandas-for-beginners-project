Crea una lista con las calificaciones de 5 estudiantes: [8.5, 9.2, 7.8, 8.9, 10].

- Calcula el promedio de las calificaciones.

```python
# Código
cal_estudiantes = [8.5,9.2,7.8,8.9,10]
print(sum(cal_estudiantes)/len(cal_estudiantes))
```

```python
cal_estudiantes = [8.5,9.2,7.8,8.9,10]
promedio = sum(cal_estudiantes) / len(cal_estudiantes)
print(promedio)
```

```python
cal_estudiantes = [8.5,9.2,7.8,8.9,10]
total = 0
for calificacion in cal_estudiantes:
    total += calificacion

promedio_for = total / len(cal_estudiantes)
print(promedio_for)
```

```python
import numpy as np
import statistics as stats

array = np.array(cal_estudiantes)
print(array.mean())

promedio_stats = stats.mean(cal_estudiantes)
print(promedio_stats)
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)
